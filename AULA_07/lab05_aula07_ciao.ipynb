```python
# ==========================================================
# LAB 05 - MEMETICO: Meta-heuristica + Busca Local
# Tema: Intensificacao do Aprendizado e Convergencia Acelerada
# ==========================================================
import numpy as np

np.random.seed(42)  # reprodutibilidade


# Funcao Rastrigin (multiplos minimos locais)
def rastrigin(x):
    return 10 * len(x) + sum(x ** 2 - 10 * np.cos(2 * np.pi * x))


def local_search_hill_climbing(solution, step_size=0.01, max_steps=20):
    """Mecanismo de Intensificacao (Aprendizado Individual / Memetica)"""
    current_sol = np.copy(solution)
    current_fit = rastrigin(current_sol)

    historico_fit = [current_fit]           # historico de fitness
    historico_sol = [np.copy(current_sol)]  # historico de posicoes

    for _ in range(max_steps):
        # TODO RESOLVIDO: gera vizinho com ruido aleatorio
        neighbor = current_sol + np.random.uniform(-step_size, step_size, size=len(solution))
        neighbor_fit = rastrigin(neighbor)

        # Aceita apenas se o vizinho for melhor (menor fitness)
        if neighbor_fit < current_fit:
            current_sol, current_fit = neighbor, neighbor_fit

        historico_fit.append(current_fit)
        historico_sol.append(np.copy(current_sol))

    return current_sol, current_fit, historico_fit, np.array(historico_sol)


# Teste da Busca Local Isolada
initial_solution = np.array([2.5, -3.1])
refined_solution, final_fit, hist_fit, hist_sol = local_search_hill_climbing(initial_solution)

print(f"[LAB 05] Solucao Inicial: {initial_solution} | Fitness: {rastrigin(initial_solution):.4f}")
print(f"[LAB 05 - SUCESSO] Solucao Refinada: {refined_solution} | Fitness: {final_fit:.4f}")

# Busca local mais intensa (mais passos e passo maior)
ref2, fit2, hist_fit2, hist_sol2 = local_search_hill_climbing(initial_solution,
                                                              step_size=0.05, max_steps=200)
print(f"[LAB 05] Intensificada (step=0.05, 200 passos): {np.round(ref2, 4)} | Fitness: {fit2:.4f}")
print(f"[LAB 05] Minimo global teorico: [0. 0.] | Fitness: {rastrigin(np.array([0.0, 0.0])):.4f}")

# ==========================================================
# Convergencia em formato de tabela (saida textual)
# ==========================================================
print("\n[LAB 05] Convergencia da busca local (step=0.01, 20 passos):")
print(" Passo | Fitness")
print("-------+---------")
for k, f in enumerate(hist_fit):
    print(f" {k:5d} | {f:7.4f}")

melhorias = sum(1 for k in range(1, len(hist_fit)) if hist_fit[k] < hist_fit[k-1])
print(f"\n[LAB 05] Vizinhos aceitos (melhorias): {melhorias} de 20 tentativas")
print(f"[LAB 05] Ganho com step=0.01: {rastrigin(initial_solution) - final_fit:.4f} "
      f"({100*(rastrigin(initial_solution) - final_fit)/rastrigin(initial_solution):.2f}%)")
print(f"[LAB 05] Ganho com step=0.05 e 200 passos: {rastrigin(initial_solution) - fit2:.4f} "
      f"({100*(rastrigin(initial_solution) - fit2)/rastrigin(initial_solution):.2f}%)")
print(f"[LAB 05] Distancia ate o minimo global (0,0): {np.linalg.norm(ref2):.4f} "
      f"-> a busca local ficou presa em um minimo local")
```
