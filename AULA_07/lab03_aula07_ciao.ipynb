```python
# ==========================================================
# LAB 03 - PSO: Inercia, Componente Cognitiva e Social
# Tema: Otimizacao Continua de Parametros e Limiares
# ==========================================================
import numpy as np

np.random.seed(42)  # reprodutibilidade


# Funcao de Custo Esferica (minimo global em x=0, y=0)
def fitness_function(position):
    return np.sum(position ** 2)


num_particles = 10
dimensions = 2
max_iter = 15

# Inicializacao de Posicoes e Velocidades
X = np.random.uniform(-5, 5, (num_particles, dimensions))
V = np.random.uniform(-1, 1, (num_particles, dimensions))
X_inicial = np.copy(X)   # guardado para comparar inicio x fim da busca

pbest_X = np.copy(X)
pbest_fitness = np.array([fitness_function(p) for p in pbest_X])

gbest_index = np.argmin(pbest_fitness)
gbest_X = np.copy(pbest_X[gbest_index])

w = 0.5    # Inercia
c1 = 1.5   # Componente Cognitiva (Individual)
c2 = 1.5   # Componente Social (Coletiva)

convergence = []          # melhor fitness global por iteracao
gbest_trajetoria = [np.copy(gbest_X)]

for t in range(max_iter):
    for i in range(num_particles):
        r1, r2 = np.random.rand(), np.random.rand()

        # TODO RESOLVIDO: equacao de atualizacao da velocidade
        V[i] = (w * V[i]) \
             + (c1 * r1 * (pbest_X[i] - X[i])) \
             + (c2 * r2 * (gbest_X - X[i]))

        # Atualizacao da posicao
        X[i] = X[i] + V[i]

        # Avaliacao de Fitness
        current_fitness = fitness_function(X[i])
        if current_fitness < pbest_fitness[i]:
            pbest_fitness[i] = current_fitness
            pbest_X[i] = np.copy(X[i])

            if current_fitness < fitness_function(gbest_X):
                gbest_X = np.copy(X[i])

    convergence.append(fitness_function(gbest_X))
    gbest_trajetoria.append(np.copy(gbest_X))
    print(f"Iteracao {t+1:2d} | gbest = [{gbest_X[0]: .6f}, {gbest_X[1]: .6f}] | "
          f"fitness = {fitness_function(gbest_X):.8f}")

print(f"\n[LAB 03 - SUCESSO] Melhor posicao encontrada pelo Enxame (gbest): {gbest_X}")
print(f"[LAB 03] Fitness final (custo): {fitness_function(gbest_X):.10f}")

# ==========================================================
# Analise textual: dispersao do enxame (inicio x fim)
# ==========================================================
disp_inicial = np.mean(np.linalg.norm(X_inicial, axis=1))
disp_final = np.mean(np.linalg.norm(X, axis=1))

print("\n[LAB 03] Convergencia (melhor custo por iteracao):")
print(" Iter | Melhor custo (gbest)")
print("------+---------------------")
for t, c in enumerate(convergence):
    print(f" {t+1:4d} | {c:.8f}")

print(f"\n[LAB 03] Distancia media das particulas ate a origem:")
print(f"         Inicio da busca: {disp_inicial:.4f}")
print(f"         Fim da busca:    {disp_final:.4f}")
print(f"[LAB 03] Reducao do custo: {convergence[0]:.6f} -> {convergence[-1]:.6f} "
      f"({convergence[0] / convergence[-1]:.0f}x menor)")
```
