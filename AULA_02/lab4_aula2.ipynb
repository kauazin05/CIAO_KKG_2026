```python
"""
============================================================
ATIVIDADE 4 — Modelagem de um Problema Real
Selecao de tarefas de desenvolvimento para uma sprint
com carga horaria limitada
============================================================

1. DESCRICAO DO PROBLEMA
No meu trabalho como analista de backend (area de Corretor/CNO,
seguros), a cada sprint chega uma lista de tarefas de desenvolvimento
- correcoes de bugs, novas funcionalidades e melhorias tecnicas -
indicadas pelo time de negocio. Cada tarefa tem um esforco estimado
(em horas) e um valor de prioridade atribuido pelo Product Owner
(quanto essa tarefa importa para o negocio). Como a equipe tem uma
capacidade de horas limitada dentro da sprint, nem todas as tarefas
cabem. O problema e: quais tarefas escolher para maximizar o valor
entregue ao negocio, sem estourar a carga horaria disponivel da equipe?

2. MODELAGEM FORMAL
- Representacao de uma solucao: vetor binario x = (x1, ..., xn),
  onde xi = 1 significa que a tarefa i foi escolhida para a sprint
  e xi = 0 que nao foi.
- Espaco de busca: cada uma das n tarefas pode ou nao ser escolhida,
  independente das demais, entao existem 2^n solucoes candidatas.
  Com n = 12 tarefas, 2^12 = 4096 combinacoes possiveis.
- Funcao objetivo: maximizar valor_total = soma(valor_i * xi)
- Restricoes: soma(horas_i * xi) <= capacidade_horas_sprint
  Uma solucao que viole essa restricao e infactivel (a equipe nao
  teria tempo de entregar tudo aquilo na sprint).

3. CLASSIFICACAO: FACIL OU DIFICIL?
Esse problema e uma instancia do Problema da Mochila 0/1 (0/1 Knapsack)
- as tarefas fazem o papel dos itens, as horas o papel do peso, e o
valor de negocio o papel do valor. E um problema classicamente
NP-dificil: nao ha algoritmo polinomial conhecido que resolva todas
as instancias, e a busca por forca bruta cresce exponencialmente (2^n).
Existe uma DP que resolve exatamente em O(n * capacidade), que e
pseudo-polinomial (rapida para capacidades pequenas, como as ~40h de
uma sprint), mas o problema continua NP-dificil no sentido formal.
Na pratica, para uma sprint real (10-20 tarefas), forca bruta ou DP
resolvem sem problema; para centenas de tarefas, a heuristica gulosa
(Atividade 3) vira necessaria.
============================================================
"""

import random

# ----------------------------------------------------------
# 1. DADOS DO PROBLEMA
# ----------------------------------------------------------
tarefas = [
    'Corrigir bug login',
    'Endpoint novo cadastro cliente',
    'Refatorar service de apolices',
    'Otimizar query relatorio',
    'Integracao API terceiros',
    'Ajuste de validacao formulario',
    'Migracao tabela historica',
    'Melhoria de logs',
    'Testes automatizados modulo X',
    'Correcao vulnerabilidade seguranca',
    'Dashboard de metricas',
    'Cache de consultas frequentes',
]

horas = [4, 16, 12, 6, 20, 3, 10, 5, 8, 6, 14, 7]      # esforco estimado (horas)
valores = [8, 9, 6, 5, 7, 4, 3, 4, 6, 10, 5, 6]        # valor/prioridade de negocio (1 a 10)

capacidade_horas_sprint = 40   # horas disponiveis da equipe na sprint
n = len(tarefas)

# ----------------------------------------------------------
# 2. TAMANHO DO ESPACO DE BUSCA
# ----------------------------------------------------------
espaco_busca = 2 ** n
print(f'Numero de tarefas: {n}')
print(f'Tamanho do espaco de busca (2^n): {espaco_busca}')

# ----------------------------------------------------------
# 3. GERAR UMA SOLUCAO ALEATORIA
# ----------------------------------------------------------
def gerar_solucao_aleatoria(n):
    """Gera um vetor binario aleatorio representando quais tarefas
    seriam colocadas na sprint (1) ou nao (0). Pode ser factivel ou nao."""
    return tuple(random.randint(0, 1) for _ in range(n))

# ----------------------------------------------------------
# 4. FUNCAO OBJETIVO
# ----------------------------------------------------------
def calcular_valor_objetivo(solucao, valores):
    """Soma o valor de negocio das tarefas selecionadas na solucao."""
    return sum(valores[i] for i in range(len(solucao)) if solucao[i] == 1)

# ----------------------------------------------------------
# 5. VERIFICACAO DE RESTRICOES
# ----------------------------------------------------------
def verificar_restricoes(solucao, horas, capacidade_horas_sprint):
    """Retorna se a solucao respeita a capacidade de horas da sprint,
    junto com o total de horas usadas."""
    horas_totais = sum(horas[i] for i in range(len(solucao)) if solucao[i] == 1)
    factivel = horas_totais <= capacidade_horas_sprint
    return factivel, horas_totais

# ----------------------------------------------------------
# 6. EXEMPLO DE EXECUCAO
# ----------------------------------------------------------
random.seed(42)
solucao = gerar_solucao_aleatoria(n)
valor_total = calcular_valor_objetivo(solucao, valores)
factivel, horas_totais = verificar_restricoes(solucao, horas, capacidade_horas_sprint)

print('\nSolucao aleatoria gerada:', solucao)
print('\nTarefas selecionadas:')
for i in range(n):
    if solucao[i] == 1:
        print(f' - {tarefas[i]} (horas: {horas[i]}, valor: {valores[i]})')

print(f'\nValor total (funcao objetivo): {valor_total}')
print(f'Horas totais utilizadas: {horas_totais} / {capacidade_horas_sprint}')
print(f'Solucao factivel (respeita restricao de horas)? {"Sim" if factivel else "Nao"}')
```
